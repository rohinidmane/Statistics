```python
# how to create a vector
import numpy as np

# Create a vector
v = np.array([2, 5]) #np.array([2, 5]) creates a 2D vector.
print("Vector v:", v)

#This vector v can be visualized as an arrow from origin (0,0) to point (2,5).
```

```python
# scaling vectors

import numpy as np

v = np.array([2, 5])
scaled_v = 3 * v

print("Original vector v:", v)
print("Scaled vector (3 × v):", scaled_v) #The vector [2, 5] scaled by 3 becomes [6, 15].

#The direction remains the same, but the magnitude increases 3 times.
```

```python
import matplotlib.pyplot as plt

origin = [0, 0]
plt.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1, color='r', label='v')
plt.quiver(*origin, *scaled_v, angles='xy', scale_units='xy', scale=1, color='g', label='3×v')
#plt.quiver() is a function in Matplotlib used to draw arrows (also called vectors) on a 2D plot.
plt.grid()
plt.xlim(0, 10)
plt.ylim(0, 20)
plt.legend()
plt.title("Vector Scaling")
plt.show()
```

```python
# vector addition

import numpy as np

a = np.array([2, 3])
b = np.array([4, 1])

sum_vec = a + b

print("Vector a:", a)
print("Vector b:", b)
print("a + b =", sum_vec)
```

```python
# vector subtraction
import numpy as np

a = np.array([5, 4])
b = np.array([2, 1])

difference = a - b

print("Vector a:", a)
print("Vector b:", b)
print("a - b =", difference)
```

```python
#dot product
import numpy as np

a = np.array([1, 3])
b = np.array([4, 2])

dot = np.dot(a, b)

print("Dot Product of a and b:", dot)
```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Projection of a onto b
proj = (np.dot(a, b) / np.dot(b, b)) * b
print("Projection of a onto b:", proj)

```

Given:
a = [1, 2, 3]
b = [4, 5, 6]

Formula for projection of a onto b:
proj_b(a) = (a · b) / (b · b) * b

Step 1: Dot product
a · b = (1×4) + (2×5) + (3×6)
      = 4 + 10 + 18
      = 32

Step 2:
b · b = (4×4) + (5×5) + (6×6)
      = 16 + 25 + 36
      = 77

Step 3: Projection vector
proj_b(a) = (32 / 77) * [4, 5, 6]
          = [128/77, 160/77, 192/77]
          ≈ [1.6623, 2.0779, 2.4935]

Final Answer:
Projection of a onto b = [1.6623, 2.0779, 2.4935]


```python
# Cross Product
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

cross = np.cross(a, b)
print("Cross product of a and b:", cross)

# Magnitude of cross product
magnitude = np.linalg.norm(cross)  # This computes the magnitude (length) or norm of the vector cross.
print("\nMagnitude of cross product (area of parallelogram):", magnitude)
```

* Vector a: [1 2 3]
* Vector b: [4 5 6]
* Cross product of a and b: [-3  6 -3]

Step-by-step computation:
* x-component: 2*6 - 3*5 = 12 - 15 = -3
* y-component: 3*4 - 1*6 = 12 - 6 = 6
* z-component: 1*5 - 2*4 = 5 - 8 = -3

* Magnitude of cross product (area of parallelogram): 7.3484692283495345


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# -------------- Vector Addition ----------------
def plot_vector_addition():
    a = np.array([2, 1])
    b = np.array([1, 3])
    c = a + b

    plt.figure()
    plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='r', label='a')
    plt.quiver(a[0], a[1], b[0], b[1], angles='xy', scale_units='xy', scale=1, color='b', label='b')
    plt.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1, color='g', label='a+b')

    plt.xlim(0, 5)
    plt.ylim(0, 5)
    plt.grid(True)
    plt.legend()
    plt.title('Vector Addition (Tip-to-Tail)')
    plt.show()

# -------------- Vector Subtraction ----------------
def plot_vector_subtraction():
    a = np.array([3, 2])
    b = np.array([1, 4])
    diff = a - b

    plt.figure()
    plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='r', label='a')
    plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='b', label='b')
    plt.quiver(b[0], b[1], diff[0], diff[1], angles='xy', scale_units='xy', scale=1, color='g', label='a-b')

    plt.xlim(-1, 5)
    plt.ylim(-1, 5)
    plt.grid(True)
    plt.legend()
    plt.title('Vector Subtraction')
    plt.show()

# -------------- Dot Product ----------------
def plot_dot_product():
    a = np.array([3, 2])
    b = np.array([2, 1])

    # Projection of a onto b
    proj_length = np.dot(a, b) / np.linalg.norm(b)
    proj_vector = (proj_length / np.linalg.norm(b)) * b

    plt.figure()
    plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='r', label='a')
    plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='b', label='b')
    plt.quiver(0, 0, proj_vector[0], proj_vector[1], angles='xy', scale_units='xy', scale=1, color='g', label='Projection of a on b')

    plt.xlim(0, 4)
    plt.ylim(0, 3.5)
    plt.grid(True)
    plt.legend()
    plt.title('Dot Product & Projection')
    plt.show()

# -------------- Cross Product ----------------
def plot_cross_product():
    a = np.array([1, 2, 0])
    b = np.array([2, 1, 0])
    cross = np.cross(a, b)

    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.quiver(0, 0, 0, a[0], a[1], a[2], color='r', label='a')
    ax.quiver(0, 0, 0, b[0], b[1], b[2], color='b', label='b')
    ax.quiver(0, 0, 0, cross[0], cross[1], cross[2], color='g', label='a × b')

    ax.set_xlim([0, 3])
    ax.set_ylim([0, 3])
    ax.set_zlim([0, 3])
    ax.legend()
    ax.set_title('Cross Product (3D Perpendicular Vector)')
    plt.show()

# ----------- Run All Plots ------------
plot_vector_addition()
plot_vector_subtraction()
plot_dot_product()
plot_cross_product()
```
